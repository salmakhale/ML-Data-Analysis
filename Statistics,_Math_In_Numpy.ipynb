## Statistics

```python
import numpy as np
```

```python
data = [10, 10, 30, 40, 50, 60, 70, 80, 90, 100]
# mean
```

```python
mean=np.mean(data)
print(mean)
```

```python
median=np.median(data)
print(median)
```

```python
from scipy import stats
mode = stats.mode(data)
print(mode)
```

```python
# range
range_value= np.ptp(data)
print(range_value)
```

```python
# variance
variance=np.var(data, ddof=0)
print(variance)
```

```python
# sd
sd=np.std(data,ddof=0)
print(sd)
```

```python
# Quartile
q=np.quantile(data, [0.25, 0.5, 0.75])
print(q)
```

```python
percentile=np.percentile(data, 25)
print(percentile)
```

```python
percentile=np.percentile(data, 90)
print(percentile)
```
