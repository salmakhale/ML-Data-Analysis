```python
num=int(input("The Numbre is :"))
factorial = 1
for i in range(1,num + 1):
       factorial = factorial*i
print("The factorial of",num,"is",factorial)
```

```python
num=int(input("Enter number: "))
rev=0
while num>0:
  rem=num%10
  rev=rev*10+rem
  num=num//10
print(rev)
```
