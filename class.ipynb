```python
from os import name
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.__age = age
        self._gender = gender

    def display_name(self):
        return f"Name: {self.name}"

    def __display_age(self):
        return f"Age: {self.__age}"

    def _display_gender(self):
        return f"Gender: {self._gender}"

    def display_all_personal_info(self):
        return f"{self.display_name()}, {self.__display_age()}, {self._display_gender()}"


class Student(Person):
    def __init__(self, name, age, gender, email, track):
        super().__init__(name, age, gender)
        self.email = email

        self._track = track

    def display_email(self):
        return f"Email: {self.email}"

    def _display_track(self):
        return f"track: {self._track}"

    def display_student_details(self):
        personal_info = self.display_all_personal_info()
        student_info = f"{self.display_email()}, {self._display_track()}"
        return f"{personal_info}, {student_info}"


student = Student('Salma Khaled', 22, "Female", "salma@gmail.com", "Informatics")

print(student.display_student_details())
```
